# NumPy Exploration

## 1. List Multiplication comparison

```python
import numpy as np
import time

# Size of the arrays
N = 1_000_000

#---Using Python lists---
list1 = list(range(N))
list2 = list(range(N))

start_time = time.time()
result = [list1[i] * list2[i] for i in range(N)]
end_time = time.time()
print("Python List Time:", end_time-start_time, "seconds")

#---Using NumPy Arrays---
array1=np.arange(N)
array2 = np.arange(N)

start_time = time.time()
result = array1 * array2 #Vectorized operation
end_time = time.time()
print("NumPy Array time:", end_time - start_time, "seconds")
```

### NumPy: Best for CPU
### TensorFlow: Best for GPU

## 2. Multiply two 100*100 matrices

```python
import random

N = 100
A = []
B = []
C = []

for i in range(N):
    A.append([0]*N)
    B.append([0]*N)    
    C.append([0]*N)
for i in range(N):
    for j in range(N):
        A[i][j] = random.random()
for i in range(N):
    for j in range(N):
        B[i][j] = random.random()
for i in range(N):
    for j in range(N):
        C[i][j] = 0

time_start = time.time()
for i in range(N):
    for j in range(N):
        for k in range(N):
            C[i][j] += A[i][k]*B[k][j]
time_end = time.time()

print("The total time taken for Python lists = ", time_end - time_start)

A1 = np.random.rand(N, N)
B1 = np.random.rand(N, N)

time_start_np = time.time()

C1 = np.matmul(A1, B1)
#Or C1 = A1 @ B1

time_end_np = time.time()
print("The total time taken with NumPy = ", time_end_np - time_start_np)
```

# NumPy mini challenge
### You are given the marks of 5 students in 3 subjects as a 2D array

### Calculate theaverage marks for each student. 
### Identify the top-performing student

```python
A = np.array([[90,95,70],[60,70,80],[85,86,87],[93,92,89],[78,69,90]])
mean_scores = np.mean(A, axis=1)
print(mean_scores)#Row-wise, 5 elements

max_mean_scores = np.where(mean_scores == max(mean_scores))[0][0]
print(max_mean_scores)

subject_top_score = np.max(A, axis = 0)# Column-wise
print(subject_top_score)# 3 elements

students_with_highest_subject_score = np.argmax(A, axis = 0)
print("Subject toppers: ", students_with_highest_subject_score)

student_with_the_highest_average_score = np.argmax(mean_scores)
print("Overall topper is student number",student_with_the_highest_average_score+1)
```

```python
scores = np.array([56, 78, 90, 45, 88, 76, 59, 92, 67, 81])

mean = scores.mean()
std = scores.std()

z = (scores - mean) / std

print("Z-score:", z)
print("Normalized mean:", z.mean())
print("Normalized standard deviation:", z.std())
```

```python
data = np.random.randint(20, 35, size=(5, 4)) #5*4 matrix

daily_avg = np.mean(data, axis = 1)
print("Daily average: ", daily_avg)

sensor_max = np.max(data, axis = 0)
print("Max temperatures measured by sensors: ", sensor_max)

max_temp = data.max()
print("Max temperature: ",max_temp)

max_pos = np.unravel_index(data.argmax(), data.shape)
print("Position of the max temperature: ", max_pos)
```
