## The usage of `*args` and `**kwargs` in Python

````python
def f(*args, **kwargs):
    print(args)
    print(kwargs)
````

````python
f(1, 2, 3, name="Tom", age=20)
````

Now:

```Python
args = (1, 2, 3)

kwargs = {
    "name": "Tom",
    "age": 20
}
```

|Syntax|Meaning|Returns|
|-|-|-|
|`*args`|Collect extra position arguments|tuple|
|`*kwargs`|Collect extra keyword arguements|dict|
