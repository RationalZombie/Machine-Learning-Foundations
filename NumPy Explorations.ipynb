# NumPy Exploration

## 1. List Multiplication comparison

```python
import numpy as np
import time

# Size of the arrays
N = 1_000_000

#---Using Python lists---
list1 = list(range(N))
list2 = list(range(N))

start_time = time.time()
result = [list1[i] * list2[i] for i in range(N)]
end_time = time.time()
print("Python List Time:", end_time-start_time, "seconds")

#---Using NumPy Arrays---
array1=np.arange(N)
array2 = np.arange(N)

start_time = time.time()
result = array1 * array2 #Vectorized operation
end_time = time.time()
print("NumPy Array time:", end_time - start_time, "seconds")
```

### NumPy: Best for CPU
### TensorFlow: Best for GPU

## 2. Multiply two 100*100 matrices

```python
import random

N = 100
A = []
B = []
C = []

for i in range(N):
    A.append([0]*N)
    B.append([0]*N)    
    C.append([0]*N)
for i in range(N):
    for j in range(N):
        A[i][j] = random.random()
for i in range(N):
    for j in range(N):
        B[i][j] = random.random()
for i in range(N):
    for j in range(N):
        C[i][j] = 0

time_start = time.time()
for i in range(N):
    for j in range(N):
        for k in range(N):
            C[i][j] += A[i][k]*B[k][j]
time_end = time.time()

print("The total time taken for Python lists = ", time_end - time_start)

A1 = np.random.rand(N, N)
B1 = np.random.rand(N, N)

time_start_np = time.time()

C1 = np.matmul(A1, B1)
#Or C1 = A1 @ B1

time_end_np = time.time()
print("The total time taken with NumPy = ", time_end_np - time_start_np)
```
