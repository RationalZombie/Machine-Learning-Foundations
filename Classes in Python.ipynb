## Inheritance

````python
class eBook(Book):
    def __init__(self, category, title, price, size_in_mb):
        super().__init__(category, title)
        self.price = price
        self.size_in_mb = size_in_mb

    def print_attributes(self):
        print("The value of title attribute is ", self.title)
        print("The value of category attribute is ", self.category)
        print("The value of price attribute is ", self.price)
        print("The value of size_in_mb attribute is ", self.size_in_mb)
````

````python
book3 = eBook(category = "Productivity", title = "Deep work", price=24, size_in_mb=2.2)
````

````python
print(book3)
print(book3.title)
````

````python
class DetailedBook(eBook):
    def __init__(self, category, title, price, size_in_mb, author, pages):
        super().__init__(category, title, price, size_in_mb)
        self.author = author
        self.pages = pages
    def print_new_attributes(self):
        print("The value of pages attribute is", self.pages)
        print("The value of author attribute is", self.author)
````

````python
book4 = DetailedBook("Productivity","Deep work", 24, 2.2, "Cal Newport", 224)
````

````python
print(book4.category)
book4.print_info()
book4.print_attributes()
book4.print_new_attributes()
````

In inheritance, if the same function appears in multiple class, the method will be overwritten with the last one.

````python
class SimpleBook:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_borrowed = False
    def borrow_book(self):
        if self.is_borrowed == False:
            print(f"Successfully borrowed {self.title}.")
            self.is_borrowed = True
        else:
            print(f"Sorry, {self.title} has already been borrowed away.")
    def return_book(self):
        self.is_borrowed = False
        print(f"Successfully returned {self.title}.")
````

````python
# Instantiate
my_book = SimpleBook("Python Crash Course", "Eric Matthes")
# First time borrowing
my_book.borrow_book()
# Second time borrowing
my_book.borrow_book()
# Return
my_book.return_book()
````

````python
class BaseBook:
    def __init__(self, title, price):
        self.title = title
        self.price = price

class AudioBook(BaseBook):
    def __init__(self, title, price, narrator, length_in_hours):
        super().__init__(title, price)
        self.narrator = narrator
        self.length_in_hours = length_in_hours
    def get_discount_price(self):
        return 0.8 * self.price
````

````python
# 创建有声书实例
audio = AudioBook("深海探秘", 50, "张三", 12.5)
print(f"书名: {audio.title}")
print(f"朗读者: {audio.narrator}")
# 验证父类属性是否成功初始化
# 验证子类属性
print(f"原价: {audio.price}")
print(f"促销价:{audio.get_discount_price()}")# 应该输出 40.0
````

````python
class PaperBook:
    def __init__(self, title):
        self.title = title
    def display_info(self):
        print(f"This is paper book {self.title}")
class EBook(PaperBook):
    def __init__(self, title, file_format):
        super().__init__(title)
        self.file_format = file_format
    def display_info(self):
        print(f"This is eBook {self.title}, with format {self.file_format}")

class EncryptedEBook(EBook):
    def __init__(self, title, file_format, password):
        super().__init__(title, file_format)
        self.password = password
    def display_info(self):
        super().display_info()
        print(f"Proceed with password {self.password}?")

# Instantiate
secret_book = EncryptedEBook("怪诞行为学", "EPUB", "admin123")

# Call display_info()
secret_book.display_info()
        
````
